# Статистическое сравнение каналов привлечения

Проверяем различие конверсий заявки в договор между Paid Search и Organic Search с помощью Z-теста для двух долей.

$$
H_0:p_1=p_2
$$
$$
H_1:p_1\neq p_2
$$


```python
from statsmodels.stats.proportion import proportions_ztest

successes = [123, 175]  # Заключённые договоры
observations = [1182, 1736]  # Всего заявок

z_stat, p_value = proportions_ztest(
    count=successes,
    nobs=observations,
    alternative='two-sided'
)

print(f"Z-statistic: {z_stat:.3f}")
print(f"P-value: {p_value:.4f}")
```

Не отвергаем гипотезу о равенстве конверсий

```python
# Исходные данные
k1, n1 = 123, 1182  # Paid Search
k2, n2 = 175, 1736  # Organic Search

# 1. Конверсии
p1 = k1 / n1
p2 = k2 / n2

# 2. Разность конверсий
diff = p1 - p2

# 3. Стандартная ошибка разности
se = (
    p1 * (1 - p1) / n1 +
    p2 * (1 - p2) / n2
) ** 0.5

# 4. Критическое значение для 95% ДИ
z_critical = 1.96

# 5. Границы доверительного интервала
ci_low = diff - z_critical * se
ci_high = diff + z_critical * se

print(f"Разница: {diff * 100:.2f} п.п.")
print(f"95% ДИ: [{ci_low * 100:.2f}; {ci_high * 100:.2f}] п.п.")
```
